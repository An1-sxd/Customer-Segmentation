# Feature Creation :

### Date features :

```python
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], format="%d-%m-%Y")
df["Dt_Customer"].dt.year
df["Dt_Customer"].dtypes
```

```python
df["year_Customer"] = df["Dt_Customer"].dt.year
df["month_Customer"] = df["Dt_Customer"].dt.month
df["day_Customer"] = df["Dt_Customer"].dt.day
```

```python
# X = X.drop("Dt_Customer", axis=1)  # i'll do it below, because i still need it to create a column
```

### Age :

```python
from datetime import datetime

df["Age"] = datetime.now().year - df["Year_Birth"]
df["Age"]
```

### Total Children :

```python
df["Total_Children"] = df["Teenhome"] + df["Kidhome"]
df["Total_Children"]
```

### Total Amount :

```python
df["Total_Amount"] = df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"]
df["Total_Amount"]
```

### Customer since :

```python
df["Customer_Since"] = (pd.Timestamp("today") - df["Dt_Customer"]).dt.days
df["Customer_Since"]
```

```python
X = df.drop(["ID", "Dt_Customer"], axis=1)
X
```

# Encoding :

```python
X_encoded = pd.get_dummies(X, columns=cat_cols, drop_first=True)
X_encoded
```

# Scaling :

```python
binary_cols = ["AcceptedCmp1", "AcceptedCmp2", "AcceptedCmp3", "AcceptedCmp4", "AcceptedCmp5", 'Complain', 'Response']

num_cols = num_cols = X.select_dtypes(include="number").columns
num_cols = num_cols.drop(binary_cols)
```

```python
num_cols
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = X_encoded.copy()
X_scaled[num_cols] = scaler.fit_transform(X_encoded[num_cols])
```

```python
X_scaled[num_cols]
```

# Model Training :

### K_means :

```python
X_scaled
```

```python
from sklearn.cluster import KMeans

n_clusters = range(1,10)
inertias = []

for i in n_clusters:
  model = KMeans(n_clusters=i)
  model.fit(X_scaled)

  inertias.append(model.inertia_)

plt.plot(n_clusters, inertias)
```

```python
model = KMeans(n_clusters=5, random_state=42, n_init=10)
model.fit(X_scaled)
```

```python
df["Cluster"] = model.predict(X_scaled)
df["Cluster"]
```

```python
df["Cluster"].value_counts()
```

```python
features = ["Age", "Income", "Total_Amount", "Recency", "NumDealsPurchases", "NumWebPurchases", "NumCatalogPurchases", "NumStorePurchases", "NumWebVisitsMonth"]
df.groupby("Cluster")[features].mean()
```

### Visualize Clusters :

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2, random_state=42)

X_pca = pca.fit_transform(X_scaled)
X_pca
```

```python
df["PCA1"], df["PCA2"] = X_pca[: ,0], X_pca[:, 1]
y_kmeans = model.labels_
```

```python
plt.scatter(df["PCA1"], df["PCA2"], c=y_kmeans)
```
